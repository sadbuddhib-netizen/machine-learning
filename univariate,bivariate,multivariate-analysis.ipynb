```python
disease_Col = ['SMOKING', 'YELLOW_FINGERS', 'ANXIETY',
       'PEER_PRESSURE', 'CHRONIC_DISEASE', 'FATIGUE', 'ALLERGY', 'WHEEZING',
       'ALCOHOL_CONSUMING', 'COUGHING', 'SHORTNESS_OF_BREATH',
       'SWALLOWING_DIFFICULTY', 'CHEST_PAIN', 'LUNG_CANCER']
```

```python
disease_Col
```

```python
df.columns
```

```python
bins = [0, 30, 40, 50, 60, 70, 100]

labels = ['0-30', '31-40', '41-50', '51-60', '61-70', '71+']

df['AGE_GROUP'] = pd.cut(df['AGE'], bins=bins, labels=labels)
```

```python
df[disease_Col] = df[disease_Col].replace({'YES': 1, 'NO': 0})
```

```python
age_disease = df.groupby('AGE_GROUP', observed=True)[disease_Col].sum()
```

```python
age_disease.sample(5)
```

```python
sns.heatmap(age_disease,annot = True,fmt ="d" ,cmap= "YlOrRd")
plt.xlabel("Condition")
plt.ylabel("Age Group")
plt.title("Condition by Age Group")
```

```python
sns.pairplot(df[['AGE','LUNG_CANCER']],hue= 'LUNG_CANCER')
plt.show()
```

```python
sns.boxplot(
    x = 'SMOKING',
    y = 'AGE',
    hue = 'LUNG_CANCER',
    data = df
)
plt.xlabel('Smoking')
plt.ylabel('Age')
plt.title('Smoking by Age')
plt.show()


```
